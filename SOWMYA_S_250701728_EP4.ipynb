```python
# EXPERIMENT 6
# Handling Missing and Inappropriate Data in a Dataset

import numpy as np
import pandas as pd

# Upload Hotel.csv and convert it into DataFrame
# Make sure to upload the 'Hotel_Dataset.csv' file to your Colab environment
# If uploaded, it will typically be found in the /content/ directory.
df = pd.read_csv('/content/Hotel_Dataset.xlsx - Hotel_Dataset.csv')

# Display the dataset
df

# Identify duplicate rows
df.duplicated()

# Display information about the DataFrame
df.info()

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Display the dataframe
df

# Find the number of rows
len(df)

# Reset the index
index = np.array(list(range(0, len(df))))
df.set_index(index, inplace=True)

# Display index
index

# Display dataframe
df

# Remove duplicate Age_Group column
df.drop(['Age_Group.1'], axis=1, inplace=True)

# Display dataframe
df
```
